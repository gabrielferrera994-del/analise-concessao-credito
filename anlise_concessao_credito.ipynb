```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico', 'Engenheiro', 'Estudante', 'Estudante'],
    'Salário': [4500, 8000, 5000, 10000, 12000, 15000, 1200, 1500],
    'Limite_Credito': [2500, 4000, 4000, 1000, 10000, 2000, 500, 250],
    'Historico_Inadimplência': ['Sem inadimplência', 'Sem inadimplência', 'Sem inadimplência', 'Com inadimplência', 'Sem inadimplência', 'Com inadimplência',
    'Sem inadimplência', 'Com inadimplência'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento', 'Solteiro', 'Solteiro', 'Solteiro'],
    'Imovel_Proprio': ['Sem Imóvel', 'Sem Imóvel', 'Sem Imóvel', 'Com Imóvel', 'Com Imóvel', 'Com Imóvel', 'Sem Imóvel', 'Sem Imóvel']
}

df = pd.DataFrame(data)
df.head(5)

print (df)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

Relacao_Salario_Credito = sns.scatterplot(
    data=df,
    x='Salário',
    y='Limite_Credito',
hue='Historico_Inadimplência', s=100
)

plt.tight_layout()
plt.show()

Relacao_Inadimplencia_Credito = sns.barplot(
    data=df,
    x='Historico_Inadimplência',
    y='Limite_Credito',
    errorbar=None
)

plt.tight_layout()
plt.show()

Relacao_Imovel_Credito = sns.boxplot(
    data=df,
    x='Imovel_Proprio',
    y='Limite_Credito',
)

plt.tight_layout()
plt.show()
```
